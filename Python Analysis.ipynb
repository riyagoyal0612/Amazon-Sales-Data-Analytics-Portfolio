Top 10 Product by Sales

```python
Product_top_10=(df.groupby('ProductName')['TotalSales'].sum().sort_values(ascending=False).head(10).reset_index())

plt.figure(figsize=(10,6))

ax=sns.barplot(data=Product_top_10,x='ProductName',y='TotalSales',hue='ProductName',palette='tab10')
for container in ax.containers:
    ax.bar_label(container,fmt='%.0f')

plt.title('Top 10 Product by Sales')
plt.xlabel('Product Name')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Monthly Sales Trend

```python

monthly_sales=df.groupby(df['OrderDate'].dt.to_period('M'))['TotalSales'].sum().reset_index()
monthly_sales['OrderDate']=monthly_sales['OrderDate'].dt.to_timestamp()
plt.figure(figsize=(14,6))
plt.plot(monthly_sales['OrderDate'],monthly_sales['TotalSales'],marker='o')
plt.title('Monthly Sale Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(monthly_sales['OrderDate'][::3],monthly_sales['OrderDate'].dt.strftime('%b-%Y')[::3],rotation=45)
plt.show()
```

Monthly order Trend

```python

monthly_order=df.groupby(df['OrderDate'].dt.to_period('M'))['OrderID'].nunique().reset_index()
monthly_order['OrderDate']=monthly_order['OrderDate'].dt.to_timestamp()
plt.figure(figsize=(14,6))
plt.plot(monthly_order['OrderDate'],monthly_order['OrderID'],marker='o')
plt.title('Monthly order Trend')
plt.xlabel('Month')
plt.ylabel('No. of Orders')
plt.xticks(monthly_order['OrderDate'][::3],monthly_order['OrderDate'].dt.strftime('%b-%Y')[::3],rotation=45)
plt.show()
```

Sales Perfromance Vary by Year

```python
Yearly_Sales=df.groupby(df['OrderDate'].dt.year)['TotalSales'].sum().reset_index()
Yearly_Sales.columns=['Year','TotalSales']
plt.figure(figsize=(8,5))
ax=sns.barplot(data=Yearly_Sales,x='Year',y='TotalSales',hue='Year',palette='tab10')
for container in ax.containers:
    ax.bar_label(container,fmt='%.0f')
plt.title('Sales Performance By Year')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

Profit Margin by Year

```python
df['Year']=df['OrderDate'].dt.year
yearly_Profit=df.groupby('Year').agg(
    Total_sales=('TotalSales','sum'),
    Total_Profit=('TotalProfit','sum'))
yearly_Profit['Profit_Margin_%']=(yearly_Profit['Total_Profit']/yearly_Profit['Total_sales'])*100
yearly_Profit
```
